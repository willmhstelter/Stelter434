```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
from scipy import stats
```

```python
#Generate 100k-point Gaussian aka generate 100 thousand points in a normal/Gaussian distribution
#Gaussian centered at value 5.0 - miu
#Gaussian's sigma value is 0.01 - Term? standard deviation ?
d = stats.norm.rvs(loc = 5.0, scale = .01, size = 100000)
```

```python
#Plot the Gaussian

#Following line sets up a figure to be viewed. The figure is composed of one graph plot on one line. 
#Changing the 1's changes the number of plots and their arrangement ?
fig, ax = plt.subplots(1,1)

#Following line describes a histogram with 50 bins. 
#Variable 'd' is the data to be plotted - the normal distribution generated in previous cell 
#density=true makes the bars appear filled in ? No - density=true normalizes such that the data represents a probability density distribution - what would an alternative look like?
ax.hist(d, 50, density=True)

#Following line sets the font size of the x and y axes to 24 ?
plt.tick_params(labelsize = 24)

#Following line sets the range of the x axis 
plt.xlim(4.95, 5.05)

#Following line creates 1000 evenly spaced values on the x axis
x = np.linspace(4.95, 5.05, 1000)

#Following line does a few things - calculates the Gaussian distribution with miu = 5 and sigma = 0.01, then overlays it on the plot of the histogram with specified visual characteristics
ax.plot(x, stats.norm.pdf(x, loc = 5.0, scale = 0.01), linewidth = 8, alpha = .7)

#Show that puppy off
plt.show()
```

```python
#Render the Gaussian into a semi-log plot 

#Following line describes a histogram with 50 bins, showing data 'd'
#density=True means the data is a normalized probability density distribution
ax = plt.hist(d,50, density=True)

#This line alters the scale of the y-axis by taking its logarithm. Now, the plot will be a semi-log plot
plt.yscale('log')

#The following line sets the font size of the marks on the axes
plt.tick_params(labelsize = 24)

#The following line sets the range of the x values of the plot
plt.xlim([4.95,5.05])

#Show it off
plt.show()
```

```python
#Plot the Gaussian on a semi-log plot with an overlaid Gaussian curve
#Following line stats up a single plot
fig, ax = plt.subplots(1, 1)

#Following line sets the plot up with 50 bins, data 'd' normalized from/to a probability density distribution 
ax.hist(d,50, density=True)

#Following line takes the logarithm of the values - semi-log plot
plt.yscale('log')

#Font size 24
plt.tick_params(labelsize = 24)

#Following line sets range of x axis
plt.xlim([4.95,5.05])

#Following line sets up 1000 points along the x axis
x = np.linspace(4.95,5.05,1000)

#Following line sets up and overlays a Gaussian curve - 
#Does this not need to be modified for the semi-log plot, or does the plt.yscale line above handle this curve as well as the data?
ax.plot(x,stats.norm.pdf(x,loc = 5., scale = 0.01),linewidth = 8,alpha = 0.7)

#Show the plot
plt.show()
```
