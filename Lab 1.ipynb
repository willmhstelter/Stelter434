```python
#Generate a sample of 100000 points from an exponential distribution centered at 0, with scale parameter 
d = stats.expon.rvs(loc=0, scale=1, size=100000)
```

```python
#Plot the exponential

#Following line sets up a figure to be viewed. The figure is composed of one graph plot on one line. 
fig, ax = plt.subplots(1,1)

#Following line describes a histogram with 1000 bins. 
#Variable 'd' is the data to be plotted - the normal distribution generated in previous cell 
ax.hist(d, 1000, density=True)

#Following line sets the font size of the x and y axes to 24 ?
plt.tick_params(labelsize = 24)

#Following line sets the range of the x axis 
plt.xlim(0, 5)

#Following line creates 100000 evenly spaced values on the x axis
x = np.linspace(0, 5, 100000)

#Following line does a few things - calculates the exponential distribution centered at 0 with sigma = 1, then overlays it on the plot of the histogram with specified visual characteristics
ax.plot(x, stats.expon.pdf(x, loc=0, scale=1), linewidth = 4, alpha = .7)

#Show that puppy off
plt.show()
```

```python
#3 Assume the distribution 'd' is produced by background noise. How likely is it that the background produces a measurement value of X, where X = 2.5?
# The probability will be (1 - the integral of the probability function from -infinity to X)
# This is the cdf function. 
resulta = stats.expon.cdf(2.5, loc=0, scale=1)
resultb =  1 - resulta
print("The percentage chance that the background showed a value of 2.5 is", resultb)

#Use the ppf function to find the sigma value of the result.
resultc = stats.norm.ppf(resulta, loc=0, scale=1)
print("This corresponds to a sigma value of", resultc)
```
